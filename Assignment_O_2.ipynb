```python
# Grocery Shopping List Program
grocery_list = []

while True:
    print("\nMenu: 1=Add Item | 2=Remove Item | 3=View List | 4=Exit")
    choice = input("Enter your choice: ")

    if choice == "1":
        item = input("Enter item to add: ")
        grocery_list.append(item)
        print(f"{item} added to the list.")

    elif choice == "2":
        item = input("Enter item to remove: ")
        if item in grocery_list:
            grocery_list.remove(item)
            print(f"{item} removed from the list.")
        else:
            print("Item not found.")

    elif choice == "3":
        print("Current Grocery List:", grocery_list)

    elif choice == "4":
        print("Exiting program...")
        break

    else:
        print("Invalid choice, try again.")
```

```python
# Dictionary of student grades
grades = {
    "Alice": 85,
    "Bob": 92,
    "Charlie": 78,
    "David": 90,
    "Emma": 88
}

# Calculate average grade
average = sum(grades.values()) / len(grades)

print("Student Grades:", grades)
print(len(grades))
print("Average Grade:", round(average, 2))
```

```python
# Word list
word_list = ["apple", "banana", "apple", "orange", "banana", "grape", "apple"]

# Dictionary for counting
word_count = {}

for word in word_list:
    word_count[word] = word_count.get(word, 0) + 1   # get() avoids KeyError

print("Word Frequency:")
for word, count in word_count.items():
    print(f"{word}: {count}")
```

```python
# Password strength checker
password = input("Enter a password: ")

if len(password) >= 8 and any(c.isdigit() for c in password) and any(c.isalpha() for c in password):
    print("Strong Password ✅")
else:
    print("Weak Password ❌ (Must be 8+ chars, include letters & digits)")
```

```python
# Voting system
candidates = ["Candidate A", "Candidate B", "Candidate C"]
votes = {name: 0 for name in candidates}

num_voters = int(input("Enter number of voters: "))

for i in range(num_voters):
    print("\nVote Options:")
    for j, candidate in enumerate(candidates, 1):
        print(f"{j}. {candidate}")

    choice = int(input("Enter your vote (1-3): "))
    if 1 <= choice <= len(candidates):
        selected = candidates[choice - 1]
        votes[selected] += 1
        print(f"Vote recorded for {selected}")
    else:
        print("Invalid choice. Vote not counted.")

# Display results
print("\nVoting Results:")
for candidate, count in votes.items():
    print(f"{candidate}: {count} votes")
```
