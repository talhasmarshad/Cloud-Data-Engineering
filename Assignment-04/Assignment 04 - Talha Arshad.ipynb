```python
# Input first matrix
m1 = int(input("Enter number of rows for first matrix (m1): "))          # Read rows of A (m1)
n1 = int(input("Enter number of columns for first matrix (n1): "))       # Read cols of A (n1)

print("Enter the first matrix values row by row (separated by space):")
matrix1 = []                                                             # Empty list to hold A
for i in range(m1):                                                      # For each row index i = 0..m1-1
    row = list(map(int, input(f"Row {i+1}: ").split()))                  # Read a line, split by spaces, convert to ints
    while len(row) != n1:                                                # Validate: must give exactly n1 numbers
        print(f"❌ Please enter exactly {n1} values.")
        row = list(map(int, input(f"Row {i+1}: ").split()))              # Re-prompt until length matches
    matrix1.append(row)                                                  # Append this validated row to A

# Input second matrix
m2 = int(input("Enter number of rows for second matrix (m2): "))         # Read rows of B (m2)
n2 = int(input("Enter number of columns for second matrix (n2): "))      # Read cols of B (n2)

# Check compatibility for multiplication
if n1 != m2:                                                             # Rule: A(m1×n1) * B(m2×n2) needs n1 == m2
    print("❌ Matrix multiplication not possible. (Columns of first ≠ Rows of second)")
else:
    print("Enter the second matrix values row by row (separated by space):")
    matrix2 = []                                                         # Empty list to hold B
    for i in range(m2):                                                  # For each row of B
        row = list(map(int, input(f"Row {i+1}: ").split()))              # Read and parse that row
        while len(row) != n2:                                            # Validate row length
            print(f"❌ Please enter exactly {n2} values.")
            row = list(map(int, input(f"Row {i+1}: ").split()))
        matrix2.append(row)                                              # Store row in B

    # Initialize result matrix with zeros
    result = [[0 for _ in range(n2)] for _ in range(m1)]                 # Make m1×n2 zero matrix for C

    # Matrix multiplication
    for i in range(m1):                                                  # For each row i of A / result C
        for j in range(n2):                                              # For each column j of B / result C
            for k in range(n1):                                          # Iterate across A’s columns / B’s rows
                result[i][j] += matrix1[i][k] * matrix2[k][j]            # Dot product: sum A[i][k]*B[k][j]

    # Print result
    print("\nResultant Matrix (Multiplication):")
    for row in result:                                                   # Pretty print row by row
        print(row)
```
