```python
def student_records(file_name):
    try:
        with open(file_name, 'r') as file:
            lines = file.readlines()
            student_data = [line.strip().split(', ') for line in lines]
            return student_data
    except FileNotFoundError:
        print(f"Error: File '{file_name}' not found.")
        return None
data = student_records("student_records.txt")
print(data)

def analyze_grades(student_data):
    total_grades = 0
    valid_students = 0
    highest = None
    lowest = None

    for name, grade in student_data:
        try:
            grade = float(grade)
            total_grades += grade
            valid_students += 1

            if highest is None or grade > highest[1]:
                highest = (name, grade)

            if lowest is None or grade < lowest[1]:
                lowest = (name, grade)

        except ValueError:
            print(f"Invalid grade for student '{name}'. Skipping again")

    if valid_students > 0:
        average_grade = total_grades / valid_students
        print(f" Grade Analysis Results:")
        print(f"Average grade: {average_grade:.2f}")
        print(f"Highest scoring student: {highest[0]} ({highest[1]})")
        print(f"Lowest scoring student: {lowest[0]} ({lowest[1]})")
    else:
        print("No valid grades found.")


def main():
    file_name = 'student_records.txt'
    student_data = student_records(file_name)
    if student_data:
        analyze_grades(student_data)


if __name__ == "__main__":
    main()
```

```python
class Vehicle:
    def __init__(self, make, model, year):
        self.make = make
        self.model = model
        self.year = year

    def display_info(self):
        print(f"Make: {self.make}, Model: {self.model}, Year: {self.year}")


class Car(Vehicle):
    def __init__(self, make, model, year, num_doors):
        super().__init__(make, model, year)
        self.num_doors = num_doors

    def display_info(self):
        super().display_info()
        print(f"Number of Doors: {self.num_doors}")


class Truck(Vehicle):
    def __init__(self, make, model, year, cargo_capacity):
        super().__init__(make, model, year)
        self.cargo_capacity = cargo_capacity

    def display_info(self):
        super().display_info()
        print(f"Capacity: {self.cargo_capacity} tons")


class Motorcycle(Vehicle):
    def __init__(self, make, model, year, drive_type):
        super().__init__(make, model, year)
        self.drive_type = drive_type

    def display_info(self):
        super().display_info()
        print(f"Drive Type: {self.drive_type}")


car = Car("City", "Corolla", 20, 2)
truck = Truck("Moto", "170", 17, 10)
motorcycle = Motorcycle("Honda", "70", 17, "None EV")

car.display_info()
truck.display_info()
motorcycle.display_info()
```

```python
# inventory.py
class Inventory:
    def __init__(self):
        # Dictionary: {book_title: quantity}
        self.books = {}

    def add_book(self, title, quantity):
        if title in self.books:
            self.books[title] += quantity
        else:
            self.books[title] = quantity
        print(f"Added {quantity} copies of '{title}'.")

    def remove_book(self, title):
        if title in self.books:
            del self.books[title]
            print(f"Book '{title}' removed from inventory.")
        else:
            print(f"Book '{title}' not found in inventory.")

    def display_inventory(self):
        print("\n📚 Current Inventory:")
        if not self.books:
            print("No books available.")
        else:
            for title, qty in self.books.items():
                print(f"• {title}: {qty} copies")
```

```python
# main.py
import inventory
import math
from datetime import datetime, timedelta

class Library:
    def __init__(self):
        self.inventory = inventory.Inventory()
        self.borrowed_books = {}  # {title: due_date}

    def borrow_book(self, title):
        if title in self.inventory.books and self.inventory.books[title] > 0:
            self.inventory.books[title] -= 1
            due_date = datetime.now() + timedelta(days=7)  # due in 7 days
            self.borrowed_books[title] = due_date
            print(f"Book '{title}' borrowed successfully. Due on {due_date.date()}")
        else:
            print(f"Book '{title}' is not available.")

    def return_book(self, title, return_date=None):
        if title in self.borrowed_books:
            self.inventory.books[title] += 1
            due_date = self.borrowed_books[title]
            if not return_date:
                return_date = datetime.now()

            # Task 3: Use math module for late fine
            late_days = (return_date - due_date).days
            fine = math.pow(late_days, 2) * 2 if late_days > 0 else 0  # fine formula
            if fine > 0:
                print(f"Returned late by {late_days} days. Fine: Rs {fine:.2f}")
            else:
                print(f"Book '{title}' returned on time.")
            del self.borrowed_books[title]
        else:
            print(f"Book '{title}' was not borrowed from this library.")

    def filter_overdue_books(self):
        # Task 4: Lambda to filter overdue books
        today = datetime.now()
        overdue = list(filter(lambda b: self.borrowed_books[b] < today, self.borrowed_books))
        return overdue

    def report_borrowed_books(self):
        # Task 5: List comprehension for report
        return [f"{title} → Due: {due.date()}" for title, due in self.borrowed_books.items()]


# ------------------- Demo -------------------
lib = Library()

# Task 1: Manage inventory
lib.inventory.add_book("Python Programming", 5)
lib.inventory.add_book("Cloud Data Engineering", 3)
lib.inventory.add_book("AI for Everyone", 2)

# Borrow & return books (Task 2)
lib.borrow_book("Python Programming")
lib.borrow_book("Cloud Data Engineering")
lib.borrow_book("AI for Everyone")

# Simulate return (some late)
lib.return_book("Python Programming", datetime.now() + timedelta(days=10))  # late return
lib.return_book("Cloud Data Engineering", datetime.now())  # on time

# Display inventory
lib.inventory.display_inventory()

# Task 4: Lambda - filter overdue books
print("\nOverdue Books:", lib.filter_overdue_books())

# Task 5: List comprehension report
print("\nBorrowed Books Report:")
for line in lib.report_borrowed_books():
    print(line)
```

```python
def perform_operations(numbers):
    total = 0
    product = 1
    for num in numbers:
        total += num
        product *= num
    average = total / len(numbers)
    return total, product, average
def main():
    numbers = [1, 2, 3, 4, 5]
    total, product, average = perform_operations(numbers)
    print("Total:", total)
    print("Product:", product)
    print("Average:", average)

if __name__ == "__main__":
    main()
```
