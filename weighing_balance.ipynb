```python
import serial, time

ser = serial.Serial('COM4', baudrate=9600, timeout=1)  # change COM and baud

# send S (stable-only) with exact CR+LF
ser.write(b'S\r\n')
resp = ser.readline().decode(errors='ignore').strip()
print("S response:", resp)

# try SI (immediate)
ser.write(b'SI\r\n')
print("SI response:", ser.readline().decode(errors='ignore').strip())
```

```python
import socket

# === configure this ===
TCP_IP = "192.200.0.10"   # IP of your RS232-to-TCP/IP converter
TCP_PORT = 4001           # change to your converter's port (often 23, 2101, or 4001)
BUFFER_SIZE = 1024

# create socket
s = socket.socket(socket.AF_INET, socket.SOCK_STREAM)
s.connect((TCP_IP, TCP_PORT))

# send 'S' (stable) with CR+LF
s.sendall(b'Q\r\n')
data = s.recv(BUFFER_SIZE).decode(errors='ignore').strip()
print("S response:", data)

# send 'SI' (immediate) with CR+LF
s.sendall(b'SI\r\n')
data = s.recv(BUFFER_SIZE).decode(errors='ignore').strip()
print("SI response:", data)

s.close()
```

```python
import serial
import re

def digital_scale():
    ComPort = "COM3"   # adjust to your port
    try:
        SerialComs = serial.Serial(ComPort, 9600, timeout=1)
    except Exception as e:
        print("Error opening COM port:", e)
        return None

    # read one line from balance
    raw_value = SerialComs.readline().decode(encoding='ascii', errors='ignore').strip()
    print("Raw:", raw_value)

    # remove unit (kg, g, spaces, CR, LF)
    clean_value = re.sub(r'[a-zA-Z\s]+', '', raw_value)  

    try:
        weight = float(clean_value)
    except ValueError:
        print("Could not convert:", clean_value)
        weight = None

    print("Weight:", weight)
    return weight

# Example usage:
if __name__ == "__main__":
    w = digital_scale()
```
