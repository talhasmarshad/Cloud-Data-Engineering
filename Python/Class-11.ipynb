```python
# Parent class
class Father:
    def __init__(self, eye_color, hair_color):
        self.eye_color = eye_color
        self.hair_color = hair_color

    def can_paint(self):
        print("Father can paint")


class Mother:
    def __init__(self, eye_color, hair_color):
        self.eye_color = eye_color
        self.hair_color = hair_color

    def can_paint(self):
        print("Mother can cook")


# Child class (multiple inheritance)
class Child(Father, Mother):
    def __init__(self, skin_color, eye_color, hair_color, body_language):
        Father.__init__(self, eye_color, hair_color)   # call Father constructor
        Mother.__init__(self, eye_color, hair_color)   # call Mother constructor
        self.skin_color = skin_color
        self.body_language = body_language

    def can_drive(self):
        print("Child can drive")

    def story_teller(self):
        print("Child can tell stories")


# Create a Child object
child_obj = Child("fair", "blue", "black", "tough")

# Access attributes and methods
print("Eye Color:", child_obj.eye_color)
print("Hair Color:", child_obj.hair_color)
print("Skin Color:", child_obj.skin_color)
print("Body Language:", child_obj.body_language)

child_obj.can_paint()      # Inherited method (due to MRO → Father first)
child_obj.can_drive()      # Child-specific
child_obj.story_teller()   # Child-specific
```

```python
from typing import overload

class Adder:
    @overload
    def add(self, x: int, y: int) -> int: ...
    
    @overload
    def add(self, x: float, y: float) -> float: ...
    
    @overload
    def add(self, x: str, y: str) -> str: ...

    def add(self, x, y):
        # Runtime type checks
        if isinstance(x, int) and isinstance(y, int):
            return x + y
        elif isinstance(x, float) and isinstance(y, float):
            return x + y
        elif isinstance(x, str) and isinstance(y, str):
            return x + y
        else:
            raise TypeError(f"Unsupported types: {type(x)} and {type(y)}")

# Usage
# Usage
adder_obj = Adder()
print(adder_obj.add(2, 3))        # int -> 5
print(adder_obj.add(2.5, 3.1))    # float -> 5.6
print(adder_obj.add("Hi ", "Talha"))  # str -> Hi Talha


# Example of error
#print(a.add(2, "World"))   # ❌ raises TypeError
```
