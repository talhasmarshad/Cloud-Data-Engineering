```python
from openpyxl import Workbook

# Create a new Excel workbook
wb = Workbook()

# Select active sheet
ws = wb.active
ws.title = "MySheet"

# Write some data
ws["A1"] = "Name"
ws["B1"] = "Age"

ws.append(["Talha", 28])
ws.append(["Ali", 32])
ws.append(["Sara", 25])

# Save the workbook
wb.save("my_data.xlsx")

print("Excel file created: my_data.xlsx")
```

```python
from openpyxl import load_workbook

# Load workbook
wb = load_workbook("my_data.xlsx")
ws = wb.active

# Print rows
for row in ws.iter_rows(values_only=True):
    print(row)
```

```python
import json
import os

file_path = "./json_file_11.json"

# Check if file exists and is not empty
if os.path.exists(file_path) and os.path.getsize(file_path) > 0:
    with open(file_path, "r") as f:
        xxx = json.load(f)
        print("Loaded JSON:", xxx)
else:
    print("⚠️ File is empty or does not exist. Writing default data...")
    # Example default data
    data = ["a", "b", "c", "d"]
    with open(file_path, "w") as f:
        json.dump(data, f, indent=4)
    print("Default JSON written:", data)
```

```python
import json

# Step 1: Read JSON file
with open("customer_29876.json", "r") as f:
    customer_29876 = json.load(f)

print("Before update:", customer_29876)

# Step 2: Modify (add a new key/value)
customer_29876["loyalty_points"] = 120

# Step 3: Write back (overwrite the file with updated data)
with open("customer_29876.json", "w") as f:
    json.dump(customer_29876, f, indent=4)   # indent=4 makes it pretty

print("After update:", customer_29876)
```

```python
import json

with open("customer_29876.json", "r") as f:
    customer_29876 = json.load(f)

print("Before update:", customer_29876)

# Overwrite some keys
customer_29876["loyalty_points"] = 200   # replace old value
customer_29876["first name"] = "Dragon"         # overwrite name
customer_29876["status"] = "Gold"        # add/overwrite

with open("customer_29876.json", "w") as f:
    json.dump(customer_29876, f, indent=4)

print("After update:", customer_29876)
```
