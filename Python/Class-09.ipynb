Function with Functions (Waiz)

```python
def factorial(n):
    if n == 0 or n == 1:   # base case
        return 1
    else:
        return n * factorial(n - 1)   # recursive call

# Test
print(factorial(5))

```

```python
# Factorial using while loop
def factorial(n):
    result = 1
    while n > 1:     # multiply until n becomes 1
        result *= n
        n -= 1
    return result

# Example usage
num = int(input("Enter a number: "))
print(f"Factorial of {num} is {factorial(num)}")

```

***Class

```python
class Cars:
    def __init__(self, color, make_year, model):  # constructor
        self.color = color
        self.model = model
        self.make_year = make_year

    def start(self):  # method
        print("It can start")

    def stop(self):  # method
        print("It can stop")


# create object (note: arguments must follow constructor order)
civic = Cars('black', 2025, 'Honda')

# access attributes
print(civic.color,civic.make_year, civic.model )      

# call methods
civic.start()
civic.stop()
```

```python
class Human:
    def __init__(self, name, age, gender):  # 3 attributes
        self.name = name
        self.age = age
        self.gender = gender

    def speak(self):  # method 1
        print(f"Hello, my name is {self.name}.")

    def walk(self):  # method 2
        print(f"{self.name} is walking at age {self.age}.")

# Create an object
person1 = Human("Ali", 25, "Male")

# Access attributes
print(person1.name)    # Ali
print(person1.age)     # 25
print(person1.gender)  # Male

# Call methods
person1.speak()  # Hello, my name is Ali.
person1.walk()   # Ali is walking.
```

```python
class BankAccount:
    def __init__(self, owner, balance=0.0):  # constructor
        self.owner = owner
        self.balance = balance

    def deposit(self, amount):  # method to deposit money
        if amount > 0:
            self.balance += amount
            print(f"${amount} deposited. New balance: ${self.balance}")
        else:
            print("Deposit amount must be positive.")

    def withdraw(self, amount):  # method to withdraw money
        if amount > self.balance:
            print("Insufficient balance!")
        elif amount <= 0:
            print("Withdrawal amount must be positive.")
        else:
            self.balance -= amount
            print(f"${amount} withdrawn. New balance: ${self.balance}")

# Create account
account1 = BankAccount("Ali", 500)

# Access attributes
print(account1.owner)    # Ali
print(account1.balance)  # 500

# Deposit money
account1.deposit(200)    # $200 deposited. New balance: $700

# Withdraw money
account1.withdraw(100)   # $100 withdrawn. New balance: $600
account1.withdraw(1000)  # Insufficient balance!

```
