Abstact

```python
from abc import ABC, abstractmethod

# Abstract Base Class
class Shape(ABC):
    def __init__(self, name="Generic Shape"):
        self.name = name

    def display(self):
        print(f"This is a {self.name}")

    @abstractmethod
    def slides(self):
        """Abstract method - must be implemented by subclasses"""
        pass

# Derived class
class Circle(Shape):
    def __init__(self):
        super().__init__("Circle")  # Call parent constructor
    
    def slides(self):
        print("Circle has no sides")   # Implementation of abstract method

# Create object
cl_obj = Circle()

# Call methods
cl_obj.display()      # from parent class
cl_obj.slides()       # from child class

# Access attribute
print(cl_obj.name)    # prints "Circle"

        
```

Polymortism


```python
class Animal:
    def speak(self):
        return "Animal Speaks"

class Dog(Animal):
    def speak(self):
        return "Woof!"

class Cat(Animal):
    def speak(self):
        return "Mew!"

# Create objects
dog = Dog()
cat = Cat()

# Call methods
print(dog.speak())   # Woof!
print(cat.speak())   # Mew!
```

```python
from abc import ABC, abstractmethod

# Abstract class
class Animal(ABC):

    @abstractmethod
    def speak(self):
        """Abstract method that must be implemented by subclasses"""
        pass

# Concrete class Dog
class Dog(Animal):
    def speak(self):
        return "Woof!"

# Concrete class Cat
class Cat(Animal):
    def speak(self):
        return "Mew!"

# Usage
dog = Dog()
cat = Cat()

print(dog.speak())   # Woof!
print(cat.speak())   # Mew!
```

```python
class Saylani:
    def __init__(self):
        self.helpline = "786"     # Public attribute
        self._batch = "Python"    # Protected attribute (by convention)
        self.__result = "786"     # Private attribute (name mangled)

# Create object
s = Saylani()

# Accessing public variable
print("Public:", s.helpline)

# Accessing protected variable (possible but not recommended)
print("Protected:", s._batch)

# Accessing private variable directly ❌ (will cause error)
# print(s.__result)   # AttributeError

# Correct way to access private variable (name mangling)
print("Private:", s._Saylani__result)
```

```python
class StudentLogin:
    # Static private attributes (class-level)
    __username = "admin"
    __password = "1234"

    # Private method for validation
    @staticmethod
    def __dbconnectivity(username, password):
        if username == StudentLogin.__username and password == StudentLogin.__password:
            return "✅ Valid User"
        else:
            return "❌ Not a Valid User"

    # Public method to validate user
    @staticmethod
    def validate_user(username, password):
        return StudentLogin.__dbconnectivity(username, password)

    # Public method to update password
    @staticmethod
    def update_password(old_password, new_password):
        if old_password == StudentLogin.__password:
            StudentLogin.__password = new_password
            return "🔑 Password Updated Successfully"
        else:
            return "❌ Old Password Incorrect"


# ---------------------------
# Usage / Testing
# ---------------------------

# 1. Validate user
print(StudentLogin.validate_user("admin", "1234"))   # ✅ Valid User
print(StudentLogin.validate_user("admin", "wrong"))  # ❌ Not a Valid User

# 2. Update password
print(StudentLogin.update_password("1234", "newpass"))  # 🔑 Updated
print(StudentLogin.update_password("wrong", "test"))    # ❌ Old Password Incorrect

# 3. Validate with new password
print(StudentLogin.validate_user("admin", "newpass"))   # ✅ Valid User
```

```python
class StudentLogin:
    def __init__(self, username, password):
        # Private attributes
        self.__username = username
        self.__password = password

    # Private method for validation
    def __dbconnectivity(self, username, password):
        if username == self.__username and password == self.__password:
            return "✅ Valid User"
        else:
            return "❌ Not a Valid User"

    # Public method to validate user
    def validate_user(self, username, password):
        return self.__dbconnectivity(username, password)

    # Public method to update password
    def update_password(self, old_password, new_password):
        if old_password == self.__password:
            self.__password = new_password
            return "🔑 Password Updated Successfully"
        else:
            return "❌ Old Password Incorrect"

    # Public method to update username
    def update_username(self, current_username, new_username, password):
        if current_username == self.__username and password == self.__password:
            self.__username = new_username
            return "👤 Username Updated Successfully"
        else:
            return "❌ Invalid Credentials - Cannot Update Username"

    # Getter for username
    def get_username(self):
        return self.__username

    # Getter for password (masked for safety)
    def get_password(self):
        return "*" * len(self.__password)


# ---------------------------
# Usage / Testing
# ---------------------------

student1 = StudentLogin("talha", "1234")

print(student1.validate_user("talha", "1234"))   # ✅ Valid User

print(student1.update_password("1234", "newpass"))  # 🔑 Password Updated
print(student1.validate_user("talha", "newpass"))   # ✅ Valid User

print(student1.update_username("talha", "syedtalha", "newpass"))  # 👤 Username Updated
print(student1.validate_user("syedtalha", "newpass"))             # ✅ Valid User

# Getter usage
print("Current Username:", student1.get_username())  # syedtalha
print("Current Password:", student1.get_password())  # ****** (masked)
```

```python
class Demo:
    # Public method
    def public_method(self):
        return "This is a Public Method"

    # Protected method (by convention: single underscore)
    def _protected_method(self):
        return "This is a Protected Method"

    # Private method (name mangling with double underscore)
    def __private_method(self):
        return "This is a Private Method"

    # Public method to access the private one
    def access_private(self):
        return self.__private_method()


# ---------- Testing ----------
demo = Demo()
print(demo.public_method())        # ✅ Works directly
print(demo._protected_method())    # ⚠️ Works but not recommended to call directly
print(demo.access_private())       # ✅ Indirect access to private
# print(demo.__private_method())   # ❌ Will throw AttributeError
```
