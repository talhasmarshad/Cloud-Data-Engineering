```python
Input_user = input("Enter string: ").upper()
vowels = 'AEIOU'
count = sum(1 for char in Input_user if char in vowels)
print("Input string:", Input_user)
print("Vowels considered:", vowels)
print("Number of vowels:", count)
```

```python
Input_user = input("Enter string: ")
count_upper = sum(1 for char in Input_user if char.isupper())
count_lower = sum(1 for char in Input_user if char.islower())
count_whitespace = sum(1 for char in Input_user if char.isspace())
count_digit = sum(1 for char in Input_user if char.isdigit())
print("Input string:", Input_user)
print("Number of uppercase letters:", count_upper)
print("Number of lowercase letters:", count_lower)
print("Number of whitespace characters:", count_whitespace)
print("Number of digits:", count_digit)
```

```python
Input_user = input("Enter string: ")
Exchanged_string = Input_user.swapcase()
print("Input string:", Input_user)
print("Exchanged case string:", Exchanged_string)
```

```python
Input_user = input("Enter string: ")
if len(Input_user) >= 2:
    new_string = Input_user[-1] + Input_user[1:-1] + Input_user[0]
else:
    new_string = Input_user
print("Original string:", Input_user)
print("String after exchanging first and last characters:", new_string)
```

```python
name = input("Enter your full name (first middle last): ")
initials = "  "
prev_space = True or False  # Assume start is like a space

for char in name:
    if prev_space and char.isalpha():
        initials += char.upper() + ". "
        prev_space = False
    elif char == " ":
        prev_space = True

print("Initials:", initials.strip())
```

```python
Input_user = input("Enter string: ")
is_palindrome = Input_user == Input_user[::-1]
print("Input string:", Input_user)
if is_palindrome:
    print("The string is a palindrome.")
else:
    print("The string is not a palindrome.")
```

```python
# Main Program
user_password = input("Enter your password: ")

if len(user_password) < 8:
    print("Password must be at least 8 characters long.")

elif not any(char.isupper() for char in user_password):
    print("Password must contain at least one uppercase letter.")

elif not any(char.islower() for char in user_password):
    print("Password must contain at least one lowercase letter.")

elif not any(char.isdigit() for char in user_password):
    print("Password must contain at least one digit.")

else:
    print("Password is valid!")
```
